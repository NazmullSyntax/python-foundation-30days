```python
"""
Day 1: Hello World
Your very first Python program.
"""

# This is a comment. Python ignores anything after #

# 1. Basic print
print("Hello, World!")

# 2. Print multiple lines
print("My name is Alex.")
print("I am learning Python.")

# 3. Print with a separator
print("Python", "is", "awesome", sep=" | ")

# 4. Print with end character (no newline)
print("This is line 1. ", end="")
print("This is line 2 on the same line.")

# 5. Print blank line
print()

# 6. Print numbers
print(2026)
print(3.14159)

# 7. Print with formatting (f-string, preview for later days)
name = "Alex"
age = 25
print(f"My name is {name} and I am {age} years old.")
```

```python
#Variables & Data Types

"""
Day 2: Variables & Data Types
Learn how to store and inspect data.
"""

# ---------- 1. Integers ----------
age = 25
year = 2026
temperature = -5
print("age =", age, "| type:", type(age))

# ---------- 2. Floats (decimals) ----------
price = 19.99
pi = 3.14159
print("price =", price, "| type:", type(price))

# ---------- 3. Strings (text) ----------
name = "Alex"
city = 'Dhaka'
multiline = """This is
a multi-line
string."""
print("name =", name, "| type:", type(name))
print(multiline)

# ---------- 4. Booleans ----------
is_student = True
has_job = False
print("is_student =", is_student, "| type:", type(is_student))

# ---------- 5. NoneType (no value yet) ----------
middle_name = None
print("middle_name =", middle_name, "| type:", type(middle_name))

# ---------- 6. Multiple assignment ----------
x, y, z = 1, 2, 3
print("x, y, z =", x, y, z)

# Same value to many variables
a = b = c = 100
print("a, b, c =", a, b, c)

# ---------- 7. Naming rules ----------
# ✅ Valid names
user_name = "Alex"
_user2 = "test"
totalAmount = 500       # camelCase (works but not pythonic)
total_amount = 500      # snake_case (PREFERRED in Python)

# ❌ Invalid (uncomment to see errors)
# 2name = "bad"     # cannot start with a number
# my-name = "bad"   # hyphens not allowed
# class = "bad"     # reserved keyword

# ---------- 8. Constants (by convention UPPERCASE) ----------
PI = 3.14159
MAX_USERS = 100
print("PI =", PI, "| MAX_USERS =", MAX_USERS)

# ---------- 9. Type conversion ----------
num_str = "42"
num_int = int(num_str)
print(num_str, "->", num_int, "| new type:", type(num_int))

float_from_int = float(10)
print(10, "->", float_from_int)

str_from_int = str(99)
print(99, "->", str_from_int, "| type:", type(str_from_int))

# Boolean conversion: 0, "", None, [], {} are Falsy. Everything else Truthy.
print(bool(0), bool(1), bool(""), bool("hi"), bool(None), bool([]))

# ---------- 10. Dynamic typing ----------
value = 10          # int
print("value:", value, type(value))
value = "ten"       # now str — Python allows this
print("value:", value, type(value))
value = 10.5        # now float
print("value:", value, type(value))
```

```python
"""
Day 3: Operators
Learn all 7 types of operators in Python.
"""

print("=" * 50)
print("PART 1: ARITHMETIC OPERATORS")
print("=" * 50)

a = 15
b = 4

print(f"a = {a}, b = {b}")
print(f"Addition        : a + b  = {a + b}")     # 19
print(f"Subtraction     : a - b  = {a - b}")     # 11
print(f"Multiplication  : a * b  = {a * b}")     # 60
print(f"Division        : a / b  = {a / b}")     # 3.75 (always float)
print(f"Floor Division  : a // b = {a // b}")    # 3
print(f"Modulus         : a % b  = {a % b}")     # 3
print(f"Exponent        : a ** b = {a ** b}")    # 50625


print("\n" + "=" * 50)
print("PART 2: COMPARISON OPERATORS")
print("=" * 50)

x = 10
y = 20

print(f"x = {x}, y = {y}")
print(f"Equal           : x == y -> {x == y}")   # False
print(f"Not equal       : x != y -> {x != y}")   # True
print(f"Greater than    : x > y  -> {x > y}")    # False
print(f"Less than       : x < y  -> {x < y}")    # True
print(f"Greater or eq   : x >= y -> {x >= y}")   # False
print(f"Less or eq      : x <= y -> {x <= y}")   # True


print("\n" + "=" * 50)
print("PART 3: LOGICAL OPERATORS")
print("=" * 50)

is_sunny = True
is_warm = False

print(f"is_sunny = {is_sunny}, is_warm = {is_warm}")
print(f"AND : is_sunny and is_warm -> {is_sunny and is_warm}")  # False
print(f"OR  : is_sunny or is_warm  -> {is_sunny or is_warm}")   # True
print(f"NOT : not is_sunny         -> {not is_sunny}")          # False

# Short-circuit demonstration
print("\nShort-circuit demo:")
print(0 and 5)   # 0  (first is falsy -> stop)
print(0 or 5)    # 5  (first is falsy -> go next)
print(3 and 5)   # 5  (both truthy -> return last)
print(3 or 5)    # 3  (first truthy -> return first)


print("\n" + "=" * 50)
print("PART 4: ASSIGNMENT OPERATORS")
print("=" * 50)

n = 10
print(f"Start: n = {n}")

n += 5
print(f"n += 5  -> {n}")   # 15

n -= 3
print(f"n -= 3  -> {n}")   # 12

n *= 2
print(f"n *= 2  -> {n}")   # 24

n /= 4
print(f"n /= 4  -> {n}")   # 6.0

n //= 2
print(f"n //= 2 -> {n}")   # 3.0

n **= 2
print(f"n **= 2 -> {n}")   # 9.0

n %= 4
print(f"n %= 4  -> {n}")   # 1.0


print("\n" + "=" * 50)
print("PART 5: BITWISE OPERATORS")
print("=" * 50)

p = 12   # binary: 1100
q = 10   # binary: 1010

print(f"p = {p} ({bin(p)}), q = {q} ({bin(q)})")
print(f"AND (&) : p & q = {p & q} ({bin(p & q)})")    # 8
print(f"OR  (|) : p | q = {p | q} ({bin(p | q)})")    # 14
print(f"XOR (^) : p ^ q = {p ^ q} ({bin(p ^ q)})")    # 6
print(f"NOT (~) : ~p    = {~p}")                       # -13
print(f"Left shift  (<<): p << 1 = {p << 1}")          # 24
print(f"Right shift (>>): p >> 1 = {p >> 1}")          # 6


print("\n" + "=" * 50)
print("PART 6: MEMBERSHIP OPERATORS")
print("=" * 50)

fruits = ["apple", "banana", "cherry"]
print(f"fruits = {fruits}")
print(f"'banana' in fruits     -> {'banana' in fruits}")       # True
print(f"'grape' in fruits      -> {'grape' in fruits}")        # False
print(f"'grape' not in fruits  -> {'grape' not in fruits}")    # True

text = "Hello Python"
print(f"'Python' in '{text}' -> {'Python' in text}")           # True
print(f"'java' in '{text}'   -> {'java' in text}")             # False


print("\n" + "=" * 50)
print("PART 7: IDENTITY OPERATORS")
print("=" * 50)

list1 = [1, 2, 3]
list2 = [1, 2, 3]
list3 = list1

print(f"list1 = {list1}")
print(f"list2 = {list2}")
print(f"list3 = list1 -> {list3}")
print(f"list1 is list2      -> {list1 is list2}")       # False
print(f"list1 is list3      -> {list1 is list3}")       # True
print(f"list1 == list2      -> {list1 == list2}")       # True
print(f"list1 is not list2  -> {list1 is not list2}")   # True


print("\n" + "=" * 50)
print("PART 8: OPERATOR PRECEDENCE")
print("=" * 50)

result = 2 + 3 * 4 ** 2 - 10 / 5
print(f"2 + 3 * 4 ** 2 - 10 / 5 = {result}")
# Step by step:
# 4 ** 2 = 16
# 3 * 16 = 48
# 10 / 5 = 2.0
# 2 + 48 - 2.0 = 48.0

print(f"(2 + 3) * ((4 ** 2) - 10) / 5 = {(2 + 3) * ((4 ** 2) - 10) / 5}")


print("\n" + "=" * 50)
print("PART 9: REAL-WORLD EXAMPLES")
print("=" * 50)

# Example 1: Even or odd check
number = 17
is_even = (number % 2 == 0)
print(f"{number} is even? {is_even}")

# Example 2: Discount calculation
price = 1200
discount_percent = 15
final_price = price - (price * discount_percent / 100)
print(f"Original: ${price} | After {discount_percent}% off: ${final_price:.2f}")

# Example 3: Age check with logical operators
age = 20
has_id = True
can_enter = age >= 18 and has_id
print(f"Age {age}, has ID {has_id} -> Can enter: {can_enter}")
```

```python
"""
Day 4: Input & Output
Learn how to take input from the user and format output beautifully.
"""

print("=" * 55)
print("PART 1: BASIC OUTPUT WITH print()")
print("=" * 55)

# 1. Simple print
print("Hello!")

# 2. Multiple values (Python adds spaces between them)
print("Name:", "Alex", "Age:", 25)

# 3. Custom separator
print("apple", "banana", "cherry", sep=", ")
print("2026", "10", "02", sep="-")

# 4. Custom end character
print("Loading", end="")
print(".", end="")
print(".", end="")
print(".")
# Output: Loading...

# 5. Escape characters
print("Line1\nLine2")            # \n = new line
print("Name:\tAlex")             # \t = tab
print("She said \"hi\"")          # \" = quote inside string
print("Path: C:\\Users\\Alex")    # \\ = single backslash


print("\n" + "=" * 55)
print("PART 2: TAKING INPUT WITH input()")
print("=" * 55)

# input() ALWAYS returns a string
name = input("Enter your name: ")
print(f"Hello, {name}!")

# If you need a number, convert it
age_str = input("Enter your age: ")
age = int(age_str)               # convert string → int
print(f"Next year you'll be {age + 1}.")

# Shorter (one line)
age2 = int(input("Enter your age again: "))
print(f"Double your age = {age2 * 2}")

# Float input
height = float(input("Enter your height in meters: "))
print(f"Your height is {height} m")

# Note: Once you call input(), the program WAITS until the user types something
# and presses Enter.


print("\n" + "=" * 55)
print("PART 3: THREE WAYS TO FORMAT STRINGS")
print("=" * 55)

name = "Alex"
age = 25
height = 1.75

# Method 1: Old style (%, like C)
print("Method 1 (%%): %s is %d years old." % (name, age))

# Method 2: str.format()
print("Method 2 (format): {} is {} years old.".format(name, age))

# Method 3: f-strings (BEST — Python 3.6+)
print(f"Method 3 (f-string): {name} is {age} years old.")


print("\n" + "=" * 55)
print("PART 4: F-STRING POWER FEATURES")
print("=" * 55)

# 1. Expressions inside {}
a = 5
b = 3
print(f"{a} + {b} = {a + b}")
print(f"{a} × {b} = {a * b}")

# 2. Function calls inside {}
print(f"Uppercase name: {name.upper()}")
print(f"Name length: {len(name)}")

# 3. Decimal places
pi = 3.14159265
print(f"Pi (2 decimals): {pi:.2f}")
print(f"Pi (4 decimals): {pi:.4f}")

# 4. Padding and alignment
print(f"|{name:<10}|")   # left-align, width 10
print(f"|{name:>10}|")   # right-align
print(f"|{name:^10}|")   # center

# 5. Numbers with commas
big = 1234567890
print(f"With commas: {big:,}")

# 6. Percentages
score = 0.856
print(f"Score: {score:.1%}")   # 85.6%

# 7. Mixed
price = 1299.5
print(f"Price: ${price:,.2f}")


print("\n" + "=" * 55)
print("PART 5: A SMALL PROFILE BUILDER")
print("=" * 55)

# Simulated user data
user_name = "Rahim"
user_age = 22
user_city = "Dhaka"
user_gpa = 3.75

print("=" * 40)
print("        STUDENT PROFILE")
print("=" * 40)
print(f"Name     : {user_name}")
print(f"Age      : {user_age}")
print(f"City     : {user_city}")
print(f"GPA      : {user_gpa:.2f}")
print("=" * 40)


print("\n" + "=" * 55)
print("PART 6: INTERACTIVE BILL CALCULATOR")
print("=" * 55)

# Comment out to skip interactive input during testing
print("(Pretend input: item=Pen, price=25.50, qty=4, tax=5)")

item = "Pen"
price = 25.50
quantity = 4
tax_percent = 5

subtotal = price * quantity
tax = subtotal * tax_percent / 100
total = subtotal + tax

print("\n" + "-" * 40)
print("            INVOICE")
print("-" * 40)
print(f"Item     : {item}")
print(f"Price    : ${price:.2f}")
print(f"Quantity : {quantity}")
print(f"Subtotal : ${subtotal:.2f}")
print(f"Tax ({tax_percent}%) : ${tax:.2f}")
print("-" * 40)
print(f"TOTAL    : ${total:.2f}")
print("-" * 40)
```
